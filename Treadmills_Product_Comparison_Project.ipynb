```python
# check for missing values in dataframes
def MissingValues():
    '''
"df.isnull": Checks for missing values (NaNs) in your DataFrame df.
- True indicates a missing value (NaN) in that cell.
- False means there is no missing value (the data is valid in that cell).
             
".sum()": Then sums the number of missing values in each column.

".sort_values(ascending = False)": Sorts these columns based on the number of missing values, with the columns having the least missing values appearing first (ascending order).
- In Pandas, the sort_values() method doesn't have a descending parameter. 
- Instead, you achieve the same effect of sorting in descending order by setting "ascending = False"

".head(15)" select top 15 columns with highest counts of null values
    '''

print(MissingValues.__doc__)

##  Using the dataframe original
display(df.head(15))

##  Sample of Missing Value on .isnull()
display(df.isnull().head(15))

##  Using .isnull().any() to know is there any False or True value in the DataFrame
print(f'How many is False and True values!')
display(df.isnull().any())

print()

display(df.isnull().sum().sort_values(ascending = True).head(15))
```

```python
##  Duplicate a backup dataframe for conversion use.
df_bkup = df.copy()

##  Converting 'df_bkup' dataframe of all Dtype'float64' to 'int32' from 'Age', 'Education', 'Usage', 'Income' and 'Miles'.

df_bkup['Age'].replace(np.nan, int(0), inplace = True)
df_bkup['Age'] = df_bkup['Age'].astype(int)

df_bkup['Education'].replace(np.nan, int(0), inplace = True)
df_bkup['Education'] = df_bkup['Education'].astype(int)

df_bkup['Usage'].replace(np.nan, int(0), inplace = True)
df_bkup['Usage'] = df_bkup['Usage'].astype(int)

df_bkup['Fitness'].replace(np.nan, int(0), inplace = True)
df_bkup['Fitness'] = df_bkup['Fitness'].astype(int)

df_bkup['Income'].replace(np.nan, int(0), inplace = True)
df_bkup['Income'] = df_bkup['Income'].astype(int)

df_bkup['Miles'].replace(np.nan, int(0), inplace = True)
df_bkup['Miles'] = df_bkup['Miles'].astype(int)

##  display the result after converted to 'int32' and show the first 20 -> 'head(20).
display(df_bkup.head(15), df_bkup.tail(15))
```
